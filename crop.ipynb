```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import pickle
```

```python
df = pd.read_csv('Crop_recommendation.csv')
df.head
```

```python
if df.isnull().sum().sum()>0:
   df= df.dropna()

print(f"dataset loaded:{df.shape[0]}records found.")
df.head()
```

```python
# remove outliners( value above limit)
for column in df.columns[:-1]:
    q1 = df[column].quantile(0.25)
    #lower 25% data
    q3 = df[column].quantile(0.75)
    #upper 75% data
    iqr = q3-q1

    #middle 50% (sahi data)
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    #filter the dataframe
    df = df[(df[column] >= lower_limit) & (df[column] <= upper_limit )]

    print(f"Data Cleared! Total rows remaining: {len(df)}")
```

```python
# question and answer ko alag karo
x = df.drop('label', axis=1)
#inputs
y = df['label']
#output

#scaling nitrogen (100) or ph(7) ko ek barabar scale me lana
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

# splitting: 80% (study->train), 20%(test)
x_train, x_test, y_train, y_test = train_test_split(x_scaled , y , test_size=0.2 , random_state=42)
```

```python
# training Random Model Forest(100 trees milkar kam karege)
model = RandomForestClassifier(n_estimators=100 , random_state=42)

#train the model
model.fit(x_train, y_train)

#accuracy check
y_pred = model.predict(x_test)
print(f"Project Accuracy:{accuracy_score (y_test, y_pred) * 100:.2f}%")

```

```python
# ek box me model or scaler ko pack karega pickel
final_assets = {'model':model, 'scaler': scaler}
with open('model.plk', 'wb') as f:
    pickle.dump(final_assets, f)
    print("Project Assest Saved Successfully!")
```
